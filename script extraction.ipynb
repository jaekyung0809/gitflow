```python
import json
from youtube_transcript_api import YouTubeTranscriptApi, NoTranscriptFound

def extract_rich_transcript(video_id: str):
    output_filename = f"{video_id}_transcript.json"
    
    try:
        ytt_api = YouTubeTranscriptApi()
        transcript_list = ytt_api.list(video_id)
        
        try:
            transcript = transcript_list.find_transcript(['ko'])
            print(f"✅ 영상({video_id})의 한국어 원본 자막을 찾았습니다.")
        except NoTranscriptFound:
            transcript = next(iter(transcript_list)).translate('ko')
            print(f"⚠️ 원본 자막이 없어 '{transcript.language}' 자막을 한국어로 번역합니다.")
        
        raw_data = transcript.fetch()
        
        refined_data = []
        for item in raw_data:
            refined_data.append({
                "text": item.text,       
                "start": item.start,     
                "duration": item.duration 
            })
            
        with open(output_filename, 'w', encoding='utf-8') as f:
            json.dump(refined_data, f, ensure_ascii=False, indent=4)
            
        print(f"🎉 추출 완료! 저장된 파일명: {output_filename}")

    except Exception as e:
        print(f"❌ 오류 발생: {e}")

# ==========================================
# 실행부
# ==========================================
if __name__ == "__main__":
    VIDEO_ID = "afe72ZYCWoA"
    
    # 함수에 영상 ID만 넣으면 알아서 해당 ID가 들어간 이름으로 파일이 생성됩니다.
    extract_rich_transcript(VIDEO_ID)
```

```python
import json
import datetime
import re
from googleapiclient.discovery import build
from youtube_transcript_api import YouTubeTranscriptApi, NoTranscriptFound


# 구글 API 키 입력
DEVELOPER_KEY = "AIzaSyCL27goouZ3ym3hASK0VfRXhtEbI3J0PLM"
YOUTUBE_API_SERVICE_NAME = "youtube"
YOUTUBE_API_VERSION = "v3"

def get_video_metadata(video_id: str):
    """YouTube Data API를 이용해 요청하신 모든 상세 메타데이터 추출"""
    youtube = build(YOUTUBE_API_SERVICE_NAME, YOUTUBE_API_VERSION, developerKey=DEVELOPER_KEY)
    
    # 요청하는 부분(part)에 snippet, statistics, contentDetails를 모두 포함합니다.
    request = youtube.videos().list(
        part="snippet,statistics,contentDetails",
        id=video_id
    )
    response = request.execute()
    
    if not response['items']:
        raise Exception("영상을 찾을 수 없거나 API 할당량을 초과했습니다.")
        
    item = response['items'][0]
    snippet = item['snippet']
    stats = item['statistics']
    content = item['contentDetails']
    
    # 요청하신 항목들을 구조에 맞게 매핑
    return {
        "video_id": item['id'],
        "title": snippet['title'],
        "channel_id": snippet['channelId'],
        "channel_title": snippet['channelTitle'],
        "category_id": snippet['categoryId'],
        "published_at": snippet['publishedAt'],
        "description": snippet['description'],
        "tags": snippet.get('tags', []),  # 태그가 없는 영상일 경우 빈 리스트 반환
        "thumbnail": snippet.get('thumbnails', {}).get('high', {}).get('url', ''), #썸네일 해상도가 'high'인 경우의 URL을 가져옴
        "view_count": int(stats.get('viewCount', 0)),
        "duration": content['duration'],
        "source_url": f"https://www.youtube.com/watch?v={video_id}",
        "extracted_at": datetime.datetime.now(datetime.timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")
    }

def extract_transcript(video_id: str):
    ytt_api = YouTubeTranscriptApi()
    try:
        transcript_list = ytt_api.list(video_id)
        
        try:
            # 💡 1순위 한국어, 2순위 영어를 찾습니다. (번역 시도 절대 안 함!)
            transcript = transcript_list.find_transcript(['ko', 'en'])
            print(f"✅ [{video_id}] '{transcript.language_code}' 원본 자막 추출 성공")
        except NoTranscriptFound:
            # 한국어도, 영어도 없으면 깔끔하게 수집 포기 에러를 발생시킵니다.
            raise Exception("NoTranscriptFound (한국어/영어 자막 모두 없음)")
        
        raw_data = transcript.fetch()
        
        # 에러 없이 i.text 로 반환
        return [{"text": i.text, "start": round(i.start, 3), "duration": round(i.duration, 3)} for i in raw_data]
    except Exception as e:
        reason = str(e)
        print(f"⚠️ [{video_id}] 자막 수집 불가: {reason[:50]}...")
        return [{"text": f"수집불가 사유: {reason}", "start": 0, "duration": 0}]

def build_final_dataset(video_id: str):
    """메타데이터와 스크립트를 결합하여 JSON 저장"""

    try:
        print(f"🚀 [{video_id}] 데이터 수집 시작...")
        
        metadata = get_video_metadata(video_id)
        transcript = extract_transcript(video_id)
        
        # + 추가 ) 출력 파일에 이름 삽입
        raw_title = metadata['title']
        # 파일명에 쓸 수 없는 문자를 공백으로 치환
        safe_title = re.sub(r'[^\w\s\-\.,!~\[\]\(\)]', '', raw_title)
        # 연속된 공백이 생겼을 경우 하나의 공백으로 깔끔하게 정리
        safe_title = re.sub(r'\s+', ' ', safe_title)
        # 100자 제한 및 앞뒤 공백 제거
        safe_title = safe_title[:100].strip()
        
        output_filename = f"{safe_title}_data.json"
        
        final_json = {
            "metadata": metadata,
            "transcript": transcript
        }
        
        with open(output_filename, 'w', encoding='utf-8') as f:
            json.dump(final_json, f, ensure_ascii=False, indent=4)
            
        print(f"✅ 수집 완료! 파일명: {output_filename}")
        print(f"📊 제목: {metadata['title']}")
        print(f"👁️ 조회수: {metadata['view_count']:,}회 / 🏷️ 태그 수: {len(metadata['tags'])}개")

    except Exception as e:
        print(f"❌ 오류 발생: {e}")

# 실행 
if __name__ == "__main__":
    TARGET_ID = "4b8ZsFszE8I"
    build_final_dataset(TARGET_ID)

''' 
# 실행부
if __name__ == "__main__":
    # 수집하고자 하는 유튜브 영상 ID 리스트
    target_videos = [
        "afe72ZYCWoA",  # Infinite Energy Generator
        "ugDkEFU6khI",  # 평평한 지구 기초
        "oNaM8OUulEI",  # 다누리호 사기
        "4b8ZsFszE8I",  # Are perpetual motion machines possible?
        "gZHMiHg83w8",  # Contrails vs Chemtrails
        "su-fmoZUkF8",  # Destroying Flat Earth
        "o6kN38CgBUM",  # 육각수의 발견
        "Jnpj3zQrOhs",  # 와디즈 사기 기업
        "SZfv30CFiKA",  # 오투스 눈운동기
        "VMV_btjRdDg",  # 오투스 플러스 리뷰
        "4xcYaw0YZB0",  # NASA의 비밀 기술
        "15V0gUXUPko"   # Free Energy Devices
    ]

    print(f"🌟 총 {len(target_videos)}개의 영상 데이터 수집 배치를 시작합니다...\n" + "="*50)

    # for문을 이용해 리스트 안의 ID를 하나씩 꺼내어 수집 함수에 전달
    for idx, video_id in enumerate(target_videos, 1):
        print(f"\n▶️ [{idx}/{len(target_videos)}] 영상 진행 중...")
        build_final_dataset(video_id)
        
    print("\n" + "="*50)
    print("🎉 모든 데이터 수집 배치가 완료되었습니다!")
    '''
```

```python
from youtube_transcript_api import YouTubeTranscriptApi

def check_subtitle_types(video_ids):
    ytt_api = YouTubeTranscriptApi()
    
    print("🔍 유튜브 자막 유형(자동/수동) 검사를 시작합니다...\n" + "="*50)
    
    for video_id in video_ids:
        try:
            # 1. 자막 리스트 가져오기
            transcript_list = ytt_api.list(video_id)
            
            print(f"✅ [{video_id}] 자막 발견:")
            
            # 2. 해당 영상에 있는 모든 자막의 유형을 검사합니다.
            for transcript in transcript_list:
                language = transcript.language
                lang_code = transcript.language_code
                
                # 💡 핵심 속성: is_generated 가 True면 AI 자동생성, False면 수동 업로드입니다.
                if transcript.is_generated:
                    print(f"   - 🤖 AI 자동 생성됨 ({language} / {lang_code})")
                else:
                    print(f"   - 🧑‍💻 직접 업로드됨 ({language} / {lang_code})")
                    
        except Exception as e:
            # 에러 메시지를 통해 왜 못 가져왔는지 원인 파악
            error_msg = str(e)
            if "TranscriptsDisabled" in error_msg or "Subtitles are disabled" in error_msg:
                reason = "자막 비활성화됨"
            elif "Could not retrieve a transcript" in error_msg or "NoTranscriptFound" in error_msg:
                reason = "자막 완전 없음 (자동생성조차 안 됨)"
            elif "TooManyRequests" in error_msg or "429" in error_msg:
                reason = "IP 차단 (유튜브 방어막 작동)"
            else:
                reason = f"접근 불가 또는 기타 오류"
                
            print(f"❌ [{video_id}] 수집 불가: {reason}")
        
        print("-" * 50)

# 실행
if __name__ == "__main__":
    # 요청하신 영상들의 ID 리스트입니다. (URL에서 v= 뒷부분)
    target_videos = [
        "afe72ZYCWoA", # Infinite Energy Generator...
        "ugDkEFU6khI", # 평평한 지구 기초...
        "oNaM8OUulEI", # 다누리호 사기...
        "4b8ZsFszE8I", # Are perpetual motion...
        "gZHMiHg83w8", # Contrails vs Chemtrails...
        "su-fmoZUkF8", # Destroying Flat Earth...
        "o6kN38CgBUM", # 육각수의 발견...
        "Jnpj3zQrOhs", # 와디즈 사기 기업...
        "SZfv30CFiKA", # 오투스 눈운동기...
        "VMV_btjRdDg", # 오투스 플러스 리뷰...
        "4xcYaw0YZB0", # NASA의 비밀 기술...
        "15V0gUXUPko"  # Free Energy Devices... (문제의 영상)
    ]
    
    check_subtitle_types(target_videos)
```
